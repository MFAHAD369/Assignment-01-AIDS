### Question 11 : Smart Student Admission System

```python
name = input("Enter student name: ")
age = int(input("Enter age: "))
matric = float(input("Enter matric percentage: "))
inter = float(input("Enter intermediate percentage: "))
test = float(input("Enter entry test score: "))
domicile = input("Do you have domicile (yes/no): ").lower()

if age < 17:
    print("Not Eligible")
    print("Reason: Age is below 17")
elif matric < 60:
    print("Not Eligible")
    print("Reason: Matric percentage is below 60")
elif inter < 60:
    print("Not Eligible")
    print("Reason: Intermediate percentage is below 60")
elif test < 50:
    print("Not Eligible")
    print("Reason: Entry test score is below 50")
elif domicile != "yes":
    print("Not Eligible")
    print("Reason: No valid domicile")
else:
    if test >= 85 and inter >= 85:
        merit = "Excellent Merit"
        scholarship = "50%"
    elif test >= 70 and inter >= 70:
        merit = "Good Merit"
        scholarship = "25%"
    else:
        merit = "Normal Merit"
        scholarship = "No Scholarship"

    
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"Intermediate: {inter}%")
    print(f"Entry Test: {test}%")
    print("Status: Eligible")
    print(f"Merit: {merit}")
    print(f"Scholarship: {scholarship}")
```

### Question 12 : Employee Performance And Promotion System

```python
name = input("Enter employee name: ")
experience = float(input("Enter years of experience: "))
salary = float(input("Enter monthly salary: "))
performance = float(input("Enter performance score (out of 100): "))
projects = int(input("Enter number of projects completed: "))
late_days = int(input("Enter number of late days: "))

if experience < 2:
    print("Not Eligible for Promotion")
    print("Reason: Experience is below 2 years")
elif performance < 70:
    print("Not Eligible for Promotion")
    print("Reason: Performance score is below 70")
elif projects < 3:
    print("Not Eligible for Promotion")
    print("Reason: Completed projects are less than 3")
elif late_days >= 10:
    print("Not Eligible for Promotion")
    print("Reason: Late days are 10 or more")
else:
    if performance >= 90 and experience >= 5:
        level = "Senior Promotion"
    elif performance >= 80 and experience >= 3:
        level = "Mid-Level Promotion"
    else:
        level = "Junior Promotion"

    if performance >= 90:
        bonus = "20%"
    elif performance >= 80:
        bonus = "15%"
    else:
        bonus = "10%"

    
    print(f"Employee: {name}")
    print(f"Experience: {experience} years")
    print(f"Performance: {performance}")
    print(f"Projects: {projects}")
    print(f"Late Days: {late_days}")
    print("Promotion Status: Eligible")
    print(f"Promotion Level: {level}")
    print(f"Bonus: {bonus}") 
```
