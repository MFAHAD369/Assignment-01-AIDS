### Question 07 : ATM Menu

```python
balance = 50000
choice = 0
while choice != 4:
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1:
        print(f"Current Balance: {balance}")
    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print(f"New Balance: {balance}")
    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))
        if amount > balance:
            print("Insufficient balance!")
        else:
            balance -= amount
            print(f"New Balance: {balance}")
    elif choice == 4:
        print("Thank you for using ATM.")
    else:
        print("Invalid choice.")
```

### Question 08 : Number Guessing Game

```python
secret_number = 25
attempts = 0
guess = 0
while guess != secret_number:
    guess = int(input("Guess: "))
    attempts += 1
    if guess > secret_number:
        print("Too High")
    elif guess < secret_number:
        print("Too Low")
    else:
        print("Correct!")
print(f"Attempts: {attempts}")
```

### Question 09 : Student Marks Entry System

```python
total_students = 0
total_passed = 0
total_failed = 0
marks = float(input("Enter marks (-1 to stop): "))
while marks != -1:
    total_students += 1
    if marks >= 50:
        print("Pass")
        total_passed += 1
    else:
        print("Fail")
        total_failed += 1
    marks = float(input("Enter marks (-1 to stop): "))
print(f"Total Students: {total_students}")
print(f"Total Passed: {total_passed}")
print(f"Total Failed: {total_failed}")
```

### Question 11 : Student Performance Management System

```python
n = int(input("Enter number of students: "))
passed = 0
failed = 0
highest = 0
lowest = 100
class_total = 0
total_percentage = 0

for i in range(n):
    name = input("Enter student name: ")
    python = int(input("Enter marks in Python: "))
    maths = int(input("Enter marks in Mathematics: "))
    english = int(input("Enter marks in English: "))

    total = python + maths + english
    percentage = total / 300 * 100

    class_total = class_total + total
    total_percentage = total_percentage + percentage

    if percentage > highest:
        highest = percentage
    if percentage < lowest:
        lowest = percentage

    print("Total Marks:", total)
    print("Percentage:", round(percentage, 2))

    if percentage >= 50 and python >= 40 and maths >= 40 and english >= 40:
        print("Result: Pass")
        passed = passed + 1
        if percentage >= 90:
            print("Grade: A+")
        elif percentage >= 80:
            print("Grade: A")
        elif percentage >= 70:
            print("Grade: B")
        elif percentage >= 60:
            print("Grade: C")
        else:
            print("Grade: D")
    else:
        print("Result: Fail")
        failed = failed + 1

average = total_percentage / n

print("Total Students:", n)
print("Passed:", passed)
print("Failed:", failed)
print("Highest Percentage:", round(highest, 2))
print("Lowest Percentage:", round(lowest, 2))
print("Class Total Marks:", class_total)
print("Class Average:", round(average, 2))
```
